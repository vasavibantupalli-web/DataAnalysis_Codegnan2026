```python
#cleaning
#EDA-4 : identify naNvalues
import pandas as pd
df = pd.DataFrame(df)
print(df)
```

```python
#Datacleaning-2 :cleaning NULL values
df.isnull().sum()
```

```python
df['Capital Gain'] = df['Capital Gain'].fillna(df['Capital Gain'].mean())
print(df)
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
print(df)
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mode())
print(df)
```

```python
df.dropna()
print(df)
```

```python
# to fill the constant value
df.fillna(0)
```

```python
df = df.replace("?",None)
```

```python
print(df.mean)
```

```python

print(df.median)
```

```python
import pandas as pd

print(df)
df['Age'] = df['Age'].ffill()
df['Age'] = df['Age'].bfill()
df['Age'] = df['Age'].interpolate()
```

```python
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)
IQR = Q3-Q1
print("Q1:",Q1)
print("Q3:",Q3)
print("IQR:",IQR)
```

```python
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)

IQR = Q3-Q1
lower = Q1 - 1.5 * IQR
UPPER = Q3 + 1.5 * IQR
iqr_outliers = df[
    (df['Age']< lower) |
    (df['Age']> UPPER)
]
print("IQR Outliers:", len(iqr_outliers))
```

```python
import pandas as pd
from scipy.stats import zscore
df['Age_zscore'] = zscore(df['Age'])

z_outliers = df[
    df['Age_zscore'].abs()>3
]
print("z-score outliers:", len(z_outliers))
```

```python
#3. z_score trimming
df_trimmed = df[df['Age_zscore'].abs() <=3]
print(df_trimmed)
```

```python
#4.winsorization
from scipy.stats.mstats import winsorize
df['Age_winsorized'] = winsorize(
    df['Age'],
    limits=[0,0.3])
print(df.head())
```
