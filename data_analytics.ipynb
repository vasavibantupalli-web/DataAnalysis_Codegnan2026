```python
import numpy as np
arr1 = np.array([
    [10,20],
    [30,40],
    [50,60]
])
arr2 = ([
    [30,60],
    [20,10],
    [20,90]
])
arr3 = np.array([
    [60,80],
    [100,90],
    [45,55]
])
result = np.vstack((arr1,arr2,arr3))
print(result)
print(result.ndim)
result = np.hstack((arr1,arr2))
print(result)
print(result.ndim)
```
