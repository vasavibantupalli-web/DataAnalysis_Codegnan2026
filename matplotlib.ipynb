```python
import matplotlib.pyplot as plt
import numpy as np
products = ["laptop","mobile","tablet","headphones","watch"]
sales = [120,180,90,150,110]
profit = [30,50,20,40,25]
```

```python
plt.plot(products,sales,marker="*")
plt.title("product vs sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.show()
```

```python
plt.plot(products,sales,profit,marker="*",label="sales,profit")
#plt.title("product vs sales")
plt.xlabel("products")
plt.ylabel("sales,profit")
plt.legend()
plt.show()
```

```python
plt.plot(products,sales,marker="o",label="sales")
plt.plot(products,profit,marker="o",label="profit")
plt.title("sales vs profit")
plt.xlabel("products")
plt.ylabel("amount")
plt.legend()
plt.show()
#line chart
```

```python
plt.barh(products,sales,label="sales")
plt.title("product vs sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.legend()
plt.show()
#column chart
```

```python
x = np.arange(len(products))
width = 0.35
plt.bar(x-width/2,sales,width,label="sales")
plt.bar(x+width/2,profit,width,label="profit")
plt.title("sales vs profit")
plt.xlabel("products")
plt.ylabel("amount")
plt.legend()
plt.show()
#clustered chart
```

```python
x = np.arange(len(products))
width = 0.35
plt.bar(x-width/2,sales,width,label="sales")
plt.bar(x-width/2,profit,width,label="profit")
plt.title("sales vs profit")
plt.xlabel("products")
plt.ylabel("amount")
plt.legend()
plt.show()
#stacked chart
```

```python
import matplotlib.pyplot as plt
months = ["jan","feb","mar","apr","may","jun"]
monthly_sales = [100,120,150,140,180,200]
plt.plot(months,monthly_sales,marker="*")
plt.title("monthly_sales")
plt.xlabel("month")
plt.ylabel("sales")
plt.show()
```

```python
import matplotlib.pyplot as plt
months = ["jan","feb","mar","apr","may","jun"]
monthly_sales = [100,120,150,140,180,200]
plt.plot(months,monthly_sales,marker="*")
plt.fill_between(months,monthly_sales)
plt.title("monthly_sales")
plt.xlabel("month")
plt.ylabel("sales")
plt.show()
```

```python
import matplotlib.pyplot as plt
products = ["laptop","mobile","tablet","watch"]
sales = [120,180,90,110]
plt.pie(
    sales,
    labels=products,
    wedgeprops={"width":0.4}
  )
plt.text(0,0,
         "total500",
         ha = "center",
         va = "center"
)
plt.show()
```

```python
import pandas as pd
data = {
    "students":["A","B","C","D","E","F","G","H","I","J"],
    "Math":[65,70,75,80,85,60,90,72,78,88],
    "python":[70,75,80,85,90,65,95,78,82,92],
    "sql":[60,68,72,78,82,55,88,70,75,85]
}
df = pd.DataFrame(data)
print(df)
```

```python
import matplotlib.pyplot as plt
plt.violinplot(df["MATHS"])
plt.title("Distribution of Math Marks")
plt.ylabel("Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt

months = [ "Jan", "Feb", "Mar", "Apr","May","Jun"]
laptop = [40, 50, 60, 70, 80 ,90]
mobile = [ 60, 70, 80, 85,100,120]
plt.stackplot(
    months,laptop,mobile,
    labels=["laptop","mobile"]
)
plt.title("laptop and mobile sales")
plt.xlabel("month")
plt.legend()
plt.show()
```
