```python
import matplotlib.pyplot as plt
import numpy as np
products = ["laptop","mobile","tablet","headphones","watch"]
sales = [120,180,90,150,110]
profit = [30,50,20,40,25]
```

```python
plt.plot(products,sales,marker="*")
plt.title("product vs sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.show()
```

```python
plt.plot(products,sales,profit,marker="*",label="sales,profit")
#plt.title("product vs sales")
plt.xlabel("products")
plt.ylabel("sales,profit")
plt.legend()
plt.show()
```

```python
plt.plot(products,sales,marker="o",label="sales")
plt.plot(products,profit,marker="o",label="profit")
plt.title("sales vs profit")
plt.xlabel("products")
plt.ylabel("amount")
plt.legend()
plt.show()
#line chart
```

```python
# contour plot( It is 3-Dimensional but represented the 2-Dimensioanal surface )
# it is a create a small numerical  values,same equal value)
# it is Graphical map represented.
# it using the meshGrid( x,y,z(z -3D to 2D))
import numpy as np
import matplotlib.pyplot as plt
x = np.arange(1, 11)
y = np.arange(1, 11)
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2
plt.contour(X, Y, Z)
plt.title("Contour Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
# 3D line plot
# it is multiple data points with lines in three-dimensional spaces
# graph contains x,y,z.
# trends,movement,
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure()

ax = fig.add_subplot(111, projection="3d")

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
z = [5, 10, 15, 20, 25]

ax.plot(x, y, z)

ax.set_title("3D Line Plot")

plt.show()
```

```python
# 3D scatter plot
# it is individual data points in three-dimensional space.
# three numerical data (ax.scatter (x,y,z)

fig = plt.figure()

ax = fig.add_subplot(111, projection="3d")

x = [1, 2, 3, 4, 5]
y = [5, 10, 15, 20, 25]
z = [2, 4, 8, 16, 32]

ax.scatter(x, y, z)

ax.set_title("3D Scatter Plot")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
plt.show()
```

```python

#3D Bar Chart

fig = plt.figure()

ax = fig.add_subplot(111, projection="3d")

x = [1, 2, 3, 4]
y = [1, 1, 1, 1]
z = [10, 20, 15, 25]

dx = [0.5, 0.5, 0.5, 0.5]
dy = [0.5, 0.5, 0.5, 0.5]
dz = z
ax.bar3d(x, y, z, dx, dy, dz)
ax.set_title("3D Bar Chart")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 3D Wireframe Plot
x = np.linspace(-5, 5, 10)
y = np.linspace(-5, 5, 10)

X, Y = np.meshgrid(x, y)
Z = X*2 + Y*2

fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")

ax.plot_wireframe(X, Y, Z)

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
#dataframe creation
data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Math": [65, 70, 75, 80, 85, 60, 90, 72, 78, 88],
    "Python": [70, 75, 80, 85, 90, 65, 95, 78, 82, 92],
    "SQL": [60, 68, 72, 78, 82, 55, 88, 70, 75, 85]
}
df = pd.DataFrame(data)
print(df)
plt.imshow(df[["Math", "Python", "SQL"]].values, cmap="BuPu")
plt.title("student marks heatmap")
plt.xlabel("Python")
plt.ylabel("student")
plt.colorbar()
```

```python
corr = df[['Math', 'Python', 'SQL']].corr()

sns.heatmap(corr, annot=True, cmap="BuPu")

plt.show()
```
